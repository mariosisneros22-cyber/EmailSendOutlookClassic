```python
import PyPDF2
```

```python
with open('FORMATOS/FORMATO1.pdf','rb') as pdf_file:
  reader = PyPDF2.PdfReader(pdf_file)
  
  total_pages = len(reader.pages)
  print(f"Total pages: {total_pages}")
  
  first_page = reader.pages[0]
  text = first_page.extract_text()
  print(text)
```
